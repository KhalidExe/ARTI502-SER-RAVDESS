# 00a — Install the CUDA build of PyTorch (Windows)
PyPI's Windows `torch` wheel is CPU-only. This replaces it with `torch==2.11.0+cu128` (RTX 5080 / sm_120).

```python
# Installs the CUDA 12.8 build of PyTorch into this kernel's environment if the CPU build is present.
# IMPORTANT: run this in a FRESH kernel (click Restart first) — Windows locks torch's DLLs once imported.
import importlib.metadata as md, subprocess, sys
v = md.version("torch")
print("python:", sys.executable)
print("installed torch:", v)
if "+cu" not in v:
    cmd = [sys.executable, "-m", "pip", "install", "--force-reinstall", "--no-deps", "torch==2.11.0+cu128",
           "--extra-index-url", "https://download.pytorch.org/whl/cu128"]
    print(" ".join(cmd))
    r = subprocess.run(cmd, capture_output=True, text=True)
    print(r.stdout[-3000:]); print(r.stderr[-3000:])
    print("installed torch now:", subprocess.run([sys.executable, "-c", "import importlib.metadata as m; print(m.version('torch'))"],
                                                capture_output=True, text=True).stdout)
    print(">>> click Restart, then open 00_env_check.ipynb and Run All")
else:
    print("CUDA build already installed - nothing to do")
```
